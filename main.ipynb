## 7. Write a function that takes a list and returns a list of the unique values.

`NOTE: You cannot use set. 🤔`

```python
numbers = [1, 2, 2, 4, 4, 5]

def unique(numbers):
    unique_numbers = []
    for num in numbers:
        if num not in unique_numbers:
            unique_numbers.append(num)
    return unique_numbers

print(unique(numbers))

```

## 8. Write a function that returns the mode of a list, i.e.: the element that appears the most times.
`NOTE: You should not use count... 🧐`

```python
numbers = [1, 2, 2, 4, 4, 4, 5]

def mode_counter(numbers):
    unique_numbers = set(numbers)

    most_common = numbers[0]
    highest_frequency = 0

    for num in unique_numbers:
        frequency = 0

        for number in numbers:
            if number == num:
                frequency += 1

        if frequency > highest_frequency:
            highest_frequency = frequency
            most_common = num

    return most_common
    
print (mode_counter(numbers))
    
    
```

```python
# This will test your function 
test_mode(mode_counter)
```

## 9. Write a function that calculates the standard deviation of a list.
`NOTE: Do not use any libraries or already built functions. 😉`

```python
def st_dev(numbers):

    total = 0

    for number in numbers:
        total += number

    mean = total / len(numbers)

    squared_differences = []

    for number in numbers:
        difference = number - mean
        squared = difference ** 2
        squared_differences.append(squared)

    total_squared = 0

    for number in squared_differences:
        total_squared += number

    variance = total_squared / (len(numbers) - 1)

    standard_deviation = variance ** 0.5

    return standard_deviation
```

```python
# This will test your function 
test_stdev(st_dev)
```
